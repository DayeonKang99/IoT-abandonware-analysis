```python
# show distribution of change against no change (changed = true) and (changed = false)
change_domains = results_df[results_df['changed'] == True].value_counts()
no_change_domains = results_df[results_df['changed'] == False].value_counts()

print(f"Number of domains with ownership change: {len(change_domains)}")
print(f"Number of domains with no ownership change: {len(no_change_domains)}")
# visualize using pie chart


labels = ['No Change', 'Change']
sizes = [len(no_change_domains), len(change_domains)]
plt.pie(sizes, labels=labels, autopct='%1.1f%%')
plt.title('Distribution of Domain Ownership Changes')
plt.show()
```
