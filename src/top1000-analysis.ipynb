```python
def read_jsonl_file(filename):
    """
    Reads a JSONL file and returns a list of Python dictionaries.
    """
    data = []
    try:
        with open(filename, 'r', encoding='utf-8') as f:
            for line in f:
                # Use json.loads() to convert each JSON string line to a Python dict
                data.append(json.loads(line.strip()))
    except FileNotFoundError:
        print(f"Error: The file '{filename}' was not found.")
    except json.JSONDecodeError as e:
        print(f"Error decoding JSON on a line: {e}")
    return data

permission_data = read_jsonl_file('v3_apps_partial.jsonl')
```

## Data flow analysis

```python
## encryption algorithm

with open('../../vsa-crypto-analysis-result.json', 'r', encoding='utf-8') as f:
    encryption_data = json.load(f)

encryption_1000 = {}
for app in encryption_data["apps"]:
    if app["app_name"] in top1000_list:
        encryption_1000[app["app_name"]] = app["encryption_algs"]
```

```python
encryption_count = {}

for app in encryption_1000:
    for enc in encryption_1000[app]:
        try:
            check = encryption_count[enc]
            encryption_count[enc] += 1
        except:
            encryption_count[enc] = 1
```

```python
count_non_rec_enc = 0
non_rec_enc = []

for app in encryption_1000:
    non_rec_flag = False
    for enc in encryption_1000[app]:
        if any(enc==alg for alg in ['DES', 'DES2', 'RC4', 'MD5', 'BLOWFISH']):
            non_rec_flag = True
        
    if non_rec_flag == True:
        count_non_rec_enc += 1
        non_rec_enc.append(app)
```

```python
import ast
import re

files = ["gp_bl_1.txt", "gp_bl_2.txt", "gp_gen.txt", "gp_local_1.txt", "gp_local_2.txt"]

endpoints = {}
for file in files:
    with open(f"../dfa-results/{file}", 'r') as f:

        current_app = None
        for line in f:
            line = line.strip()

            if not line or line.startswith('-'):
                continue

            # Detect app name
            if not line.startswith("The flow") and not line.startswith("We reconstructed"):
                current_app = line
                continue

            # Extract reconstructed values (source or sink)
            if line.startswith("We reconstructed for the"):
                match = re.search(r"\{([^}]*)\}", line)
                if match and current_app:
                    values = re.findall(r"'([^']*)'", match.group(1))

                    endpoints.setdefault(current_app, set()).update(values)

# Convert to lists
endpoints = {k: sorted(list(v)) for k, v in endpoints.items()}
```

```python
blocklist_domains = []
high_impact_blocklist = []
with open('../dfa-results/url_blocklist_results.csv', mode='r', newline='', encoding='utf-8') as csv_file:
    csv_reader = csv.reader(csv_file, delimiter=',')
    next(csv_reader)
    for row in csv_reader:
        if row[6] == "True":
            blocklist_domains.append(row)
            if row[7] in ("malware", "spyware", "phishing"):
                high_impact_blocklist.append(row)

blocklist_domains_host = []
for url in blocklist_domains:
    blocklist_domains_host.append(url[3])

high_blocklist_domains_host = []
for url in high_impact_blocklist:
    high_blocklist_domains_host.append(url[3])
```

```python
from urllib.parse import urlparse
import tldextract

dfa_result = {}
for app in top_1000:
    dfa_result[app[0]] = {"owner_changed":[], "blocklist":[], "blocklist_high":[], "unreachable":[]}
    tmp_key = f"{app[0]}.json"
    unreachable_fqdns = unreachable_data[tmp_key]["unreachable_fqdn"]

    if not app[0] in endpoints:
        continue
    for url in endpoints[app[0]]:
        hostname = urlparse(url).hostname  # live.chartboost.com

        try:
            host_fqdn = tldextract.extract(hostname).top_domain_under_public_suffix #registered_domain #".".join(tldextract.extract(hostname)[1:3])
        except:
            print(url)

        if any(item == host_fqdn for item in ownership_changed_domains):
            dfa_result[app[0]]["owner_changed"].append(url)
        if any(item == hostname for item in high_blocklist_domains_host):
            dfa_result[app[0]]["blocklist_high"].append(url)
        if any(item == hostname for item in blocklist_domains_host):
            dfa_result[app[0]]["blocklist"].append(url)
        if any(item == hostname for item in unreachable_fqdns):
            dfa_result[app[0]]["unreachable"].append(url)
```

```python
count_total = []
count_high_impact = []

for app in dfa_result:
    record = dfa_result[app]
    if any(record.values()):
        count_total.append(app)
    if any(record[k] for k in ['owner_changed', 'blocklist_high', 'unreachable']):
        count_high_impact.append(app)
```

```python
print(f"count:{len(count_total)} - {count_total}")
print(f"count:{len(count_high_impact)} - {count_high_impact}")
```

```python
deprecated_total = []
deprecated_high_impact = []

for app in non_rec_enc: 
    if app in count_total:
        deprecated_total.append(app)
    if app in count_high_impact:
        deprecated_high_impact.append(app)
```

```python
print(f"count:{len(deprecated_total)} - {deprecated_total}")
print(f"count:{len(deprecated_high_impact)} - {deprecated_high_impact}")
```
